# Objects
There are various data types in python such as string, integer, float, list, tuple, dictionary and Boolean. These are **objects** in Python. An object is characterised by:

- a data type
- an internal representation
- set of functions called **methods** used to interact with the data

An object is an instance of a particular data type, such as 1 is an instance of <code>type int</code>, [1,2,3] is an instance of <code>type list</code>, etc.  

# Classes 

It is a set of functions that a data type provides to interact with the type. AN example is functions such as <code>sorted</code> associated with lists for sorting them, here lists are the object and sorted the method. A class has 

- Data attributes or parameters
- Methods

A class on rectangles would contain data attributes such aas length and width and methods. A class is created using the <code>class</code> keyword
  

## Constructor Method

The <code> __ init __</code> method is a special method known as the **constructor**. It initializes the instance attributes (also called instance variables) when an object is created. 

The <code>self</code> parameter.  is the first parameter within the constructor for referencing to the instance being created.


```python
# A Simple Class

class Cars:  #naming the class

    def __init__(self, mileage, max_speed, year):
        self.mileage = mileage
        self.max_speed = max_speed
        self.year = year

    def car_speed(self):
        print("what is the car's max speed?")
        print(f"It is {self.max_speed} kmph")

    def car_props(self):
        print(f"the car has {self.mileage} kmpl mileage ")
        print(f"The car model is from the year {self.year}")



My_car = Cars(20, 200, 2000)   #calling the class
My_car.car_speed()          #calling the method
My_car.car_props()

print("")

My_car2 = Cars(15, 250, 2004) # class can be called multiple times
My_car2.car_speed()
My_car2.car_props()
```

```python
#Another class

import matplotlib.pyplot as plt

class Circle(object):
    
    # Constructor
    def __init__(self, radius = 3, color = 'blue'):
        self.radius = radius
        self.color = color 
    
    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)
    
    # Method 
    def drawCircle(self): 
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()  

redcircle = Circle( 10, 'darkred')
redcircle.add_radius(1)  #radius of the circle in plot is 11
redcircle.drawCircle()
```
