## Structured Query Language (SQL)

Databases refers to repositories of data. A Database Management System (DBMS) refers to the software tools used to manage Databases. A Relational Database organises data in tables, such as in spreadsheets, allowing relationaships within different tables. Relational DBMS (or RDMBS) controls access, organisation and storage of databases, examples are MySQL, Oracle DataBase, etc. SQL is used to manage and query data within relational databases. It has the following basic commands:

- `CREATE` : generates new tables
- `INSERT` : adding new data
- `SELECT` : retrieves data from databases.
- `UPDATE` : modifying data
- `DELETE` : removing data

#### **Note : SQL commands above are not case sensitive**

`SELECT` statements is used to retrieve data from the relational database. Syntax:

- `Select * from table_name` to extract the entire database
- `Select column_name_1 , Colummn_name_2 from table_name` to extract specific columns by referring to the column heading

### Conditionals in SQL
`WHERE` clause filters data based on specific criteria. Syntax:

<code>SELECT column1 FROM table_name
 WHERE condition</code>

Various comparison operators can be used in the `WHERE` clause, such as Equal to (=), Greater than (>), Less than (<), Greater than or equal to (>=), Less than or equal to (<=), and Not equal to (!=).

### Built-in functions 

`COUNT` : Built in function that retreives the number of rows satisfying a given criteria. Syntax:

<code> SELECT COUNT(column_name) FROM table_name
    WHERE column_name = 'country' </code>

`DISTINCT` Chooses distinct elements within a column by specifying the column name

`LIMIT` funtion restricts the dta retrieval to the first n rows of the table
