```python
import pandas as pd
import sqlite3
import csv
import prettytable
prettytable.DEFAULT = 'DEFAULT'
%load_ext sql
%sql sqlite:///final.db
conn = sqlite3.connect("final.db")
cursor = conn.cursor()
```

```python
df = pd.read_csv('chicagocrimedata.csv')
#df = pd.read_csv("chicagocrimedata.csv")
df.to_sql("chicago_crime_data", conn, if_exists='replace', index=False,method="multi")

```

```python
df = pd.read_csv("chicagopublicschools.csv")
df.to_sql("chicago_public_schools" , conn , if_exists = 'replace' , index = False)
```

```python
df1 = pd.read_csv('chicagocensusdata.csv')
df1.to_sql("census_data" , conn , if_exists = 'replace' , index = False, method = "multi")
```
