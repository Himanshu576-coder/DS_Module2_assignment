# Dictionary

- Built in data type consisting of **Key** - **Value** pairs, analogous to index-element pairs in lists
- Keys are immutable and unique while values are mutable and can be duplicated
- Enclosed within {} braces, with key separated by : followed by value
- Keys and Values are stored in dictionaries as lists
  

Keys and values can of any data type but keys are **immutable** though values aren't

```python
ds_course = {"module1": "What is DS", "module2": "Tools in DS", "module3": "DS methodology",
        "module4": "Python for DSAI", 'module5': "Project" , "module6" : "Database and SQL" ,
        "module7" : "data analysis" , "module8" : "visualisation" , "module9": "ML in Python" , 
        "module10" : "Capstone and GenAI" }

print(ds_course["module10"]) #accessing key value

print(ds_course.keys() ,'\n \t' ,Dict.values()) # observe that keys and values are stored as lists separately
print("")

ds_course["module11"] = "career prep" #adding key-value pair in dict
print(ds_course["module11"])
```

## Additional Methods
- <code>dict_name.items()</code> retrieves all key-value pairs as tuples and converts them into a list of tuples. Each tuple consists of a key and its corresponding value.
- to create a copy of a dict, use <code>dict_name.copy()</code> , which creates a shallow copy of the dictionary. The new dictionary contains the same key-value pairs as the original, but they remain distinct objects in memory.
- use <code>dict_name.update({key:value})</code> to update a key-value pair
- The <code>dict_name.clear()</code> method empties the dictionary, removing all key-value pairs within it. After this operation, the dictionary is still accessible and can be used further.
- <code>del dict_name[key]</code> deletes a key-value from dict
