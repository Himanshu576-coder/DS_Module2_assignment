# Functions 
SImilar to the mathematical definition of function, they take in arguments and return a value. There are two types of functions, **pre defined** functions such as print(), range(), len(), min(), max(), etc and **user defined** functions. 

*   Functions blocks begin <code>def</code> followed by the function <code>name</code> and parentheses <code>()</code>.
*   There are input parameters or arguments that should be placed within these parentheses.
*   You can also define parameters inside these parentheses.
*   There is a body within every function that starts with a colon (<code>:</code>) and is indented.
*   You can also place documentation before the body.
*   The statement <code>return</code> exits a function, optionally passing back a value.

```python
import timeit

# Efficient O(n) method
def product_except_self_efficient(nums):
    n = len(nums)
    result = [1] * n
    prefix = 1
    for i in range(n):
        result[i] = prefix
        prefix *= nums[i]
    suffix = 1
    for i in range(n - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]
    return result

# Naive O(n²) method
def product_except_self_naive(nums):
    output = []
    for i in range(len(nums)):
        val = nums.pop(i)
        prod = 1
        for num in nums:
            prod *= num
        output.append(prod)
        nums.insert(i, val)
    return output

# Sample input
arr = list(range(1, 100))  # Adjust size for clearer difference

# Efficient
efficient_time = timeit.timeit(lambda: product_except_self_efficient(arr), number=100)

# Naive
naive_time = timeit.timeit(lambda: product_except_self_naive(arr), number=100)

print(f"Efficient method time: {efficient_time:.4f} seconds")
print(f"Naive method time: {naive_time:.4f} seconds")
```
