**Problem**: Write code to check if the number is a palindrome and if not, then reverse the number and add it to the original until it becomes a palindrome.

```python
# Palindrome test
n = 1095

def check_palindrome():
    n_str = str(n)
    rev_n = n_str[-1::-1]
    if rev_n == n_str:
        return n_str + " is palindrome"
    else:
        #return "n is not palindrome"
        while True:
            n_str = int(n_str) + int(rev_n)
            newn_str = str(n_str)
            rev_newn = newn_str[-1::-1]
            if rev_newn == newn_str:
                return newn_str + " is a palindrome"
                break
    
check_palindrome()

    
```
