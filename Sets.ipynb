# Sets 
- Sets are unordered data structures enclosed within {} used for storing unique information and performing set operations
- <code> .add() </code> adds elements one at a time, akin to append in lists, similarly <code>.remove()</code> removes elements one at a time
- <code>.pop()</code> removes an arbitrary element from the set. 
- 

```python
A = { 1 , 2 , "Krishna" , "Mira" }
A.add("Radha") #single arguments
print(A) # see that order of elements changes
A.pop()
print(A)

L = list(A) + ["typecast"]
print(L)
```

#### Typecasting 
It is the process of converting a data structure to another. If a set needs to be converted to list, write <code>list_name = list(set_name)</code> 
#### Set Operations
- Union = use <code> A | B </code> or <code>A.union(B)</code> the + operator is not recognised for sets
- Intersection = use ampersand & to denote intersection as <code> A & B </code> or <code>A.intersection(B)</code>
- Contained = use <code>A.issubset(B)</code>, use <code>issuperset</code> to check of superset
- Difference = use <code>A.difference(B)</code>
- Symmetric difference = use <code>A.symmetric_difference(B)</code>

```python
A = {1, 2, 3}
B = {2, 3, 5}
print(A.intersection(B))
print(A.difference(B))
```
