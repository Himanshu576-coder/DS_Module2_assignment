## String enumeration and slicing

Character enumerating in strings starts from zero. Last character can be obtained using negative numbers, like -1 is assigned to last element, -2 to second last and so on.

To **slice** string elements at certain indices, write `str_name[1:4:2]` which extracts elements of str_name from index 1 to 4 not including 4, taking a jump of step 2, ie starts at element 1 and then to element 3. By default, it starts at 0 and ends when the string terminates with a jump of 1

```python
#string manipulation
name = "Radha Rani"
print(name[1]) #calling an element from string

print(name[-1], name[-2])

```

```python
name = 'Radha rani'
print(name[0:5:2]) #note the element at index 5 isn't printed
print(name[5:1:-1])
print(len(name)) 
```

## String methods

**Strings are immutable** that is if any methods are applied to a string variable, the original variable is unchanged and only a copy is created. So whenever applying string methods store the new string within a new variable

<code> name.upper() </code> capitalises everything

<code> name.replace("Radha", "Shyam") </code> replace the first segment by the second

<code> name.find("Radha") </code> finds the given segment, and if not found returns -1

`name.strip(" KRISHNA  ")` removes spaces in strings



# Supplementary stuff

In Python, RegEx (short for Regular Expression) is a tool for matching and handling strings. Python provides a built-in module called <code>re</code>, which allows you to work with regular expressions. This RegEx module provides several functions for working with regular expressions, including <code>search, split, findall,</code> and <code>sub</code>. 

First, import the <code>re</code> module. The search() function searches for specified patterns within a string. Here is an example that explains how to use the search() function to search for the word "Body" in the string "The BodyGuard is the best".

```python
import re
s1 = "The BodyGuard is the best album"

# Define the pattern to search for
pattern = r"Body"

# Use the search() function to search for the pattern in the string
result = re.search(pattern, s1)

# Check if a match was found
if result:
    print("Match found!")
else:
    print("Match not found.")
```

```python
import re
#pattern = r"\d\d\d\d\d\d\d\d\d\d"  # Matches any ten consecutive digits
pattern = r"123"
text = "My Phone number is 1234567890"
match = re.search(pattern, text)

if match:
    print("Phone number found:", match.group()) 
#The match.group() method is used in Python's re module to retrieve the part of the string where the 
#regular expression pattern matched. 
else:
    print("No match")
```

Problem: Write code to check if the number is a palindrome and if not, then reverse the number and add it to the original until it becomes a palindrome.

```python
# Palindrome test
n = 19

def check_palindrome():
    n_str = str(n)
    rev_n = n_str[-1::-1]
    if rev_n == n_str:
        return n_str + " is palindrome"
    else:
        #return "n is not palindrome"
        while True:
            n_str = int(n_str) + int(rev_n)
            newn_str = str(n_str)
            rev_newn = newn_str[-1::-1]
            if rev_newn == newn_str:
                return newn_str + " is a palindrome"
                break
    
check_palindrome()
```
