# Working with BeautifulSoup
BeautifulSoup helps in working with HTML and XML documents. It helps in parsing HTML documents by utilising a tree structure. An HTML doc has the structure `<head>` and `<body>` with parent and child subcategories or tags. 

- `soup.prettify()` organises the html
- `soup.get_text()` extracts the text from the page 
- 

## Objects in BeautifulSoup

There are 4 types of objects: `Tag`, `NavigableString`, `BeautifulSoup`, and `Comment`

- `Tag`  represents a single HTML/XML tag (e.g., `<div>`, `<a>`, `<p>`). Tags have a number of methods and attributes associated with them, such as:
  - `name`: the name of the tag can be accessed as `tag.name`. It can be changed
  - `attrs`: refers to attributes, which are the values associated to tags, and work very similar to dictionaries. An attribute can be accessed by treating the tag like a key (by using `tag.attrs.keys()`), and attributes can also be modified similarly. Multi valued attributes are stored as lists
  -
