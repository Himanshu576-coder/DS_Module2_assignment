# Accessing and opening files 

## Opening files
Files can be opened using Python's <code> open() </code> function, which creates a file object and accesses the data within a **text** file. It takes two arguments:
- the file path
- file mode, "r" for reading, "w" for writing and "a" for appending

Using the <code> with </code> statement closes the file after execution of the indented block, so that <code> close() </code> method is not needed. *Read* raises an error if a file isn't found at the given location, while *write* and *append* will create a file in case the file isn't found

<code>.read(n) </code> reads n characters at a time, and in case no argument is passed then reads the entire document. On the other hand <code>readline(n) </code> reads a certain line upto n characters, and reads the full line if no arguments are passed.

```python
with open('Untitled Folder/jsk.txt','r') as file:
    print(file.readline(2)) # reads out the first line till 20 characters. then starts a new line if characters are left
    print(file.read(20))     
    for i in range(6):
        print(file.readline())
        
   
```

## Writing and Appending files
<code> 'w' </code> command is used to write in files. <code> file_name.write( "argument") </code> is used to write to the file, but note that **write() command will remove previous data and overwrite existing contents**. To avoid this, append command <code>'a' </code> must be used.

```python
with open('Untitled Folder/jsk.txt' , 'w') as file:
    file.write("Haribol \n")
    file.write("jai shri krishna \njai shri radhe")
    krishn = ["Govind" , "Murali Manohar"]
    for line in krishn:
        file.write("\n")
        file.write(line)
    
```

```python
with open('jsk.txt' , 'w') as file:
    file.write("krishna \n")
    file.write("radha \n")

with open('jsk.txt' , 'a') as file:
    file.write('this is appending')

```

## Cursor control in file
Opening the file in **w** is akin to opening the .txt file, moving your cursor to the beginning of the text file, writing new text and deleting everything that follows.
Whereas opening the file in **a** is similiar to opening the .txt file, moving your cursor to the very end and then adding the new pieces of text. <br>
It is often very useful to know where the 'cursor' is in a file and be able to control it. The following methods allow us to do precisely this -
- <code>.tell()</code> - returns the current position in bytes
- <code>.seek(offset,from)</code> - changes the position by 'offset' bytes with respect to 'from'. From can take the value of 0,1,2 corresponding to beginning, relative to current position and end


```python
with open('jsk.txt', 'r') as file:
    print(file.tell())

with open('jsk.txt', 'a+') as file1:
    for i in range (2):
        file1.write(f'\n {i} 1')
    contents = file1.read()
    print(contents)


          
```

```python
with open('Untitled folder/newjsk.txt' , 'w') as file:
    file.write('this is the new file meant for copying')
    file.seek(1 , 0)
    file.write('a')
    file.seek(2 , 0)
    file.write('b')
    i = 0
    file.seek(20 , 0)
    for i in range(6) :
        file.write(f"\n this is line {i} ")
        i = i + 1
    
```

## Copying Files

```python
with open('Untitled folder/jsk.txt' , 'r') as sourcefile:
    with open('Untitled folder/newjsk.txt' , 'w') as copyfile:
        copyfile.seek(10 ,0)
        for i in sourcefile:
            copyfile.write(i) 
```
