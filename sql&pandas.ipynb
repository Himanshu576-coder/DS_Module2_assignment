```python
import pandas as pd
import sqlite3
import csv
import prettytable
prettytable.DEFAULT = 'DEFAULT'
%load_ext sql
%sql sqlite:///socioeconomic.db
conn = sqlite3.connect("socioeconomic.db")
cursor = conn.cursor()
```

### Store the dataset in a Table

##### In many cases the dataset to be analyzed is available as a .CSV (comma separated values) file, perhaps on the internet. To analyze the data using SQL, it first needs to be stored in the database.

##### We will first read the csv files  from the given url  into pandas dataframes

##### Next we will be using the  df.to_sql() function to convert each csv file  to a table in sqlite  with the csv data loaded in it.


```python
df = pd.read_csv('https://data.cityofchicago.org/resource/jcxq-k9xf.csv')
df.to_sql("chicago_socioeconomic_data", conn, if_exists='replace', index=False,method="multi")
```

```python
%%sql select count(hardship_index) from chicago_socioeconomic_data where hardship_index > 50;

```
