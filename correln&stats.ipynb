# Correlation
Correlation measures the interdependency of variables and how thy impact each other such as rain and umbrellas are correlated. 

## Pearson's correlation

measures strength of correlation betwen two features. It gives two values, correlation coefficient and p-value. 
- **correlation coefficient**: Values close to +1 indicate large positive relationship, close to -1 indicates large negative relationship and close to 0 indicates no relation.
  
- **p-Value**: indicates certainity about the correlation measured.
    - p-value $\leq$ 0.001 indicates strong certainity
    - Moderate certainity for p-Value between 0.001 and 0.05
    - Weak ceratinity for p-values between 0.05 and 0.1
    - p-value $\geq$ 0.1 indicates no certainity

## Continuous variables and relations between them
Continuous variables can be understood using scatter plots with the fitted line. In order to start understanding the (linear) relationship between an individual variable and the price, we can use **`regplot`** which plots the scatterplot plus the fitted regression line for the data. This will be useful later on for visualizing the fit of the simple linear regression model as well.

```python
sns.regplot(x="engine-size", y="price", data=df)
plt.ylim(0,)
df[['engine-size' , 'price']].corr()
#positive linear relationship 
```

```python
sns.regplot(x = 'highway-mpg' , y = 'price' , data = df)
#plt.ylim(0,)
df[['highway-mpg' , 'price']].corr()
#Negative correlation
```

```python
sns.regplot(x = 'peak-rpm' , y = 'price' , data = df)
#weak correlation
```

### Pearson correlation coefficient & p value using scipy stats package

```python
pearson_corr_coeff , p_value = stats.pearsonr(df['price'] , df['highway-mpg'])
print(pearson_corr_coeff , p_value)
if p_value < 0.001 :
    print("strong correlation")
```

```python
pearson_corr_coeff , p_value = stats.pearsonr(df['price'] , df['peak-rpm'])
print(pearson_corr_coeff , p_value)
if p_value < 0.001 :
    print("strong correlation")
```
