# Exceptions and Errors in Python

**Exception** is an error that occurs during the execution of code. This error causes the code to raise an exception and if not prepared to handle it will halt the execution of the code.. It could be a mistake in the code or a situation that was not planned for. Python can raise these alerts automatically, but we can also trigger them on purpose using the <code>raise</code> command. We can prevent our program from crashing by handling exceptions.

While **Errors** lead to program crashes or abnormal termination, **Exceptions** are caused due to some problem within the code and can be resolved accordingly using try-except blocks.

Examples of errors are SyntaxError or NameError (occurs when variable is not defined/assigned) or and examples of exceptions are ZeroDivisionError , ArithmeticError, etc


```python
import math
math.sqrt(-2)
```

**Try and Except** : You can use the <code>try and except</code> blocks to prevent your program from crashing due to exceptions.

Here's how they work:

1. The code that may result in an exception is contained in the try block.
2. If an exception occurs, the code directly jumps to except block.
3. In the except block, you can define how to handle the exception gracefully, like displaying an error message or taking alternative actions.
4. After the except block, the program continues executing the remaining code.

```python
try:
    print(10/0)
except ZeroDivisionError:
    print("error cannot divide by zero")
```

<code>try except else finally</code> block executes as above with the <code>else</code> block executes if there are no exceptions raised in the try block. <code>finally</code> block executes irrespective of any exceptions and usually signifies the end of the <code> try except </code> blocks

## Example of ValueError

```python
import math #to use sqrt function
def sq_root(a):
    try:
        b = math.sqrt(a)
        return b
    except ValueError :
        print("please enter a positive integer")
    else:
        print(b)
    finally:
        print('process done')

sq_root(float(input("enter a number")))
        
```
